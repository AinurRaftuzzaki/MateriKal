#**Matriks**#

##**Pengenalan Matriks**##

Matriks adalah susunan bilangan-bilangan, simbol, atau ekspresi matematika yang disusun dalam bentuk persegi panjang. Susunan ini diatur dalam baris dan kolom, dan biasanya dibatasi oleh tanda kurung (baik kurung biasa "()" atau kurung siku "[]"). Setiap bilangan atau elemen dalam matriks disebut sebagai **elemen matriks** atau **unsur matriks**.

- **Baris** adalah susunan bilangan-bilangan yang disusun secara mendatar (horizontal). Misalnya, dalam matriks berordo \( 2 \times 3 \), terdapat 2 baris yang masing-masing terdiri dari 3 elemen.
- **Kolom** adalah susunan bilangan-bilangan yang disusun secara tegak (vertikal). Pada matriks yang sama, terdapat 3 kolom yang masing-masing terdiri dari 2 elemen.

**Ordo matriks** adalah ukuran matriks yang dinyatakan dalam bentuk \( m \times n \), di mana:
- \( m \) adalah banyaknya baris,
- \( n \) adalah banyaknya kolom.

Contoh:
$$ A = \begin{pmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{pmatrix} $$

Matriks \( A \) di atas memiliki:
- 2 baris (baris pertama: 1, 2, 3; baris kedua: 4, 5, 6),
- 3 kolom (kolom pertama: 1, 4; kolom kedua: 2, 5; kolom ketiga: 3, 6),
- sehingga ordo matriks \( A \) adalah \( 2 \times 3 \).

Matriks memiliki peran penting dalam berbagai bidang, seperti matematika, fisika, ilmu komputer, dan ekonomi. Mereka digunakan untuk menyelesaikan sistem persamaan linear, melakukan transformasi geometri, menganalisis data, dan banyak lagi. Dengan memahami struktur dan sifat-sifat matriks, kita dapat memanfaatkannya untuk memecahkan berbagai masalah secara efisien.

## **Jenis-jenis Matriks**##
### **1. Matriks Nol**###
Matriks yang seluruh elemennya adalah bilangan nol.

   $$ A = \begin{pmatrix}
   0 & 0 & 0\\
   0 & 0 & 0
   \end{pmatrix}  -> Matriks Nol $$


### **2. Matriks Baris**###
Matriks yang hanya memiliki satu baris, berordo 1 x j.

   $$ B = \begin{pmatrix}
   1 & 2 & -1\\
   \end{pmatrix}  -> Matriks Baris $$

### **3. Matriks Kolom**###
Matriks yang hanya memiliki satu kolom, berordo i x 1.
   
   $$ C = \begin{pmatrix}
   2\\
   3
   \end{pmatrix}  -> Matriks Kolom $$

### **4. Matriks Persegi**###
Matriks yang banyaknya baris sama dengan banyaknya kolom, berordo i x i.

   $$ D = \begin{pmatrix}
   1 & 3 \\
   -2 & 5
   \end{pmatrix}  -> Matriks Persegi $$
   
### **5. Matriks Diagonal**###
Matriks persegi yang semua elemennya nol, kecuali pada diagonal utamanya.

   $$ E = \begin{pmatrix}
   2 & 0 & 0\\
   0 & 7 & 0\\
   0 & 0 & 3 \\
   \end{pmatrix}  -> Matriks Diagonal $$

### **6. Matriks Segitiga Atas**###
Matriks persegi yang semua elemen di bawah diagonal utamanya adalah nol.

   $$ F = \begin{pmatrix}
   4 & 7 & 8\\
   0 & 2 & 1\\
   0 & 0 & 3 \\
   \end{pmatrix}  -> Matriks Segitiga Atas $$

### **7. Matriks Segitiga Bawah**###
Matriks persegi yang semua elemen di atas diagonal utamanya adalah nol.

   $$ G = \begin{pmatrix}
   4 & 0 & 0\\
   9 & 2 & 0\\
   2 & 4 & 3 \\
   \end{pmatrix}  -> Matriks Segitiga Bawah $$

### **8. Matriks Identitas**###
Matriks persegi yang elemen pada diagonal utamanya adalah satu, sedangkan elemen lainnya adalah nol.

   $$ I = \begin{pmatrix}
   1 & 0 & 0\\
   0 & 1 & 0\\
   0 & 0 & 1 \\
   \end{pmatrix}  -> Matriks Identitas $$

## **Operasi Matriks**##
Berikut beberapa operasi matriks:
###1. **Penjumlahan Matriks**###
Dua matriks dapat dijumlahkan jika memiliki ordo yang sama.

Contoh:
   
   $$ \begin{pmatrix}
   1 & 2 \\
   3 & 4
   \end{pmatrix}
   +
   \begin{pmatrix}
   5 & 6 \\
   7 & 8
   \end{pmatrix}
   =
   \begin{pmatrix}
   6 & 8 \\
   10 & 12
   \end{pmatrix} $$


###2. **Perkalian Matriks**###
Perkalian dua matriks \( A \) dan \( B \) dapat dilakukan jika jumlah kolom matriks \( A \) sama dengan jumlah baris matriks \( B \).

Contoh:

   $$ \begin{pmatrix}
   1 & 2 \\
   3 & 4
   \end{pmatrix}
   \times
   \begin{pmatrix}
   5 & 6 \\
   7 & 8
   \end{pmatrix}
   =
   \begin{pmatrix}
   19 & 22 \\
   43 & 50
   \end{pmatrix} $$

###3. **Transpose Matriks**###
Transpose matriks \( A \) (ditulis \( A^T \)) adalah matriks yang baris dan kolomnya dipertukarkan.

Contoh:
   
   $$ A = \begin{pmatrix}
   1 & 2 \\
   3 & 4
   \end{pmatrix}, \quad
   A^T = \begin{pmatrix}
   1 & 3 \\
   2 & 4
   \end{pmatrix} $$
   

###4. **Determinan**###
Determinan adalah nilai skalar yang dapat dihitung dari matriks persegi.

Contoh:
   
   $$ \text{det}(A) = \begin{vmatrix}
   1 & 2 \\
   3 & 4
   \end{vmatrix}
   = (1 \cdot 4) - (2 \cdot 3) = -2 $$
   

###5. **Invers Matriks**###
Invers matriks \( A \) (ditulis \( A^{-1} \)) adalah matriks yang jika dikalikan dengan \( A \) menghasilkan matriks identitas.

Contoh:
   
   $$ A = \begin{pmatrix}
   1 & 2 \\
   3 & 4
   \end{pmatrix}, \quad
   A^{-1} = \begin{pmatrix}
   -2 & 1 \\
   1.5 & -0.5
   \end{pmatrix} $$

## **Contoh Soal**##
1. Diberikan matriks:
   
   $$ A = \begin{pmatrix}
   2 & 3 \\
   1 & 4
   \end{pmatrix}, \quad
   B = \begin{pmatrix}
   5 & 6 \\
   7 & 8
   \end{pmatrix} $$
   Hitung \( A + B \) dan \( A x B \).

   **Jawaban:**
   $$ A + B = \begin{pmatrix}
   7 & 9 \\
   8 & 12
   \end{pmatrix}, \quad
   A \times B = \begin{pmatrix}
   31 & 36 \\
   33 & 38
   \end{pmatrix} $$

2. Tentukan determinan dari matriks:
   
   $$ C = \begin{pmatrix}
   3 & 1 \\
   2 & 4
   \end{pmatrix} $$

   **Jawaban:**
  $$ \text{det}(C) = (3 \cdot 4) - (1 \cdot 2) = 10 $$
